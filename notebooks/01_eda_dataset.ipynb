```python
# Usamos este script para comprobar que el dataset se ha descargado correctamente y que podemos acceder a él.
import json

with open("../data/raw/v1.1/mvtec_screws.json") as f:
    data = json.load(f)

print(data.keys())
```

```python
# Imprimimos las categorías, la primera imagen y la primera anotación para comprobar que todo está correcto.
print(data["categories"])
print()
print(data["images"][0])
print()
print(data["annotations"][0])
```

```python
# Contammos el número de instancias por categoría para ver si el dataset está balanceado o no.
from collections import Counter
import pandas as pd

cat_names = {cat["id"]: cat["name"] for cat in data["categories"]}

counts = Counter([ann["category_id"] for ann in data["annotations"]])
df_counts = pd.DataFrame(
    [(cat_names[cid], n) for cid, n in counts.items()],
    columns=["categoria", "n_instancias"]
).sort_values("n_instancias", ascending=False)

print(df_counts)
print("\nTotal anotaciones:", sum(counts.values()))
```

Buenas noticias: el dataset está balanceado; podemos trabajar directamente sin aplicar técnicas de balanceo, lo que simplifica bastante las cosas.
